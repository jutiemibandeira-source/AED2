```python
'''Projeto 1: Encontrando seu Filme Favorito 
Contexto: Imagine um sistema de recomendação de filmes simples onde os filmes são armazenados em uma 
lista não-ordenada. 
Problema: Implementar uma função que permita ao usuário buscar um filme pelo título nessa lista.
Crie uma lista de títulos de filmes e uma função que use a busca sequencial para 
encontrar um título inserido pelo usuário. Se o filme for encontrado, exiba uma mensagem informando o título e 
sua posição (índice) na lista. Se não for encontrado, exiba uma mensagem informando que o filme não está no 
catálogo.'''

def buscar_filmes(titulo, lista_filmes):
    for i in range(len(lista_filmes)):
        if lista_filmes[i] == titulo:
            return i 
    return -1

lista_filmes = ["JURASSIC PARK", "DE VOLTA PARA O FUTURO", "TITANIC", "VELOZES E FURIOSOS", "HOMEM ARANHA", "A ODISSÉIA"]

filme_escolhido = input("Digite o título do filme que deseja buscar: ").upper().strip()
indice = buscar_filmes(filme_escolhido, lista_filmes)

if indice != -1:
    print(f"Filme '{filme_escolhido}' encontrado na posição {indice}.")

else: 
    print(f"O filme {filme_escolhido} não está no catálogo")
```

```python
'''Projeto 2: Verificando a Lista de Convidados 
Contexto: Desenvolver um programa para gerenciar a lista de convidados de um evento. 
Problema: Implementar uma funcionalidade que permita verificar se um nome específico está na lista de 
convidados. 
Indicação de Solução: Crie uma lista de nomes de convidados. Implemente uma função que receba um nome 
como entrada e use a busca sequencial para verificar se esse nome existe na lista. Exiba uma mensagem 
indicando se o convidado está na lista ou não.'''

def busca_convidados(convidado, lista_convidados):
    for i in range(len(lista_convidados)):
        if lista_convidados[i] == convidado:
            return i
    return -1

lista_convidados = ["Julia", "Ana", "Marcelo", "Anderson","Felipe","Lisandra", "Lorena","Bruna"]
nome = input("Digite o nome do convidado: ").capitalize().strip()
indice = busca_convidados(nome, lista_convidados)

if indice != -1: 
    print(f"O nome '{nome}' foi encontrado na posição {indice} da lista de convidados.") 

else: print(f"O nome '{nome}' não foi encontrado na lista.")
```
