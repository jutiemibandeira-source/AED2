Exercício 4 — Classe Livro: atributos e métodos  [Fácil] 
Crie uma classe Livro para representar um livro de uma biblioteca. 
Requisitos: 
• O construtor deve receber titulo, autor e paginas. 
• Crie o método descricao(), que retorne uma string no formato: “Título — Autor (N páginas)”. 
• Instancie dois livros diferentes e exiba suas descrições. 

```python
class Livro:
    def __init__ (self, titulo, autor, paginas):
        self.titulo = titulo
        self.autor = autor
        self.paginas = paginas

    def descricao(self):
        return f"Título: {self.titulo} - Autor: {self.autor} - Páginas: {self.paginas}"

l1 = Livro('Crepúsculo', 'Stephenie Meyer',498)
l2 = Livro('Harry Potter e a Pedra Filosofal', 'J.K. Rowling', 264)

print(l1.descricao())
print(l2.descricao())
```

```python
class Livro:
    def __init__ (self, titulo, autor, paginas):
        self.titulo = titulo
        self.autor = autor
        self.paginas = paginas

    def __str__(self):
        return f"Título: {self.titulo} - Autor: {self.autor} - Páginas: {self.paginas}"

l1 = Livro('Crepúsculo', "Stephenie Meyer",498)
l2 = Livro('Harry Potter e a Pedra Filosofal', 'J.K. Rowling', 264)

print(l1)
print(l2)
```
