Exercício 5 — Conta simples com encapsulamento  [Fácil] 

Crie uma classe Conta que mantenha o saldo protegido contra alteração direta. 
Requisitos: 
• O construtor recebe o nome do titular e inicia o saldo em 0. 
• Use __saldo como atributo privado. 
• Implemente depositar(valor), permitindo apenas valores positivos. 
• Implemente consultar_saldo(), retornando o saldo atual. 
• Crie uma conta, faça dois depósitos e mostre o saldo. 

```python
class Conta:
    def __init__ (self,titular):
        self.titular = titular
        self._saldo = 0

    def depositar(self,valor):
        if valor < 0:
            return "Tente colocar um saldo positivo"
        else:
            self._saldo += valor 
        return self._saldo

    def consultar_saldo(self):
        return f"O valor presente em sua conta é {self._saldo}"

c1 = Conta("Julia")
print(c1.depositar(100))
c1.consultar_saldo()

    
```

Exercício 6 — Animais: herança e polimorfismo  [Fácil] Implemente uma pequena hierarquia de animais para exercitar herança e polimorfismo. 
 Requisitos: 
 • Crie a classe Animal com atributo nome e método emitir_som(). 
 • Crie Cachorro e Gato herdando de Animal. 
 • Redefina emitir_som(): Cachorro retorna “au au” e Gato retorna “miau”. 
 • Crie uma lista com um cachorro e um gato e percorra-a chamando emitir_som() sem testar o tipo do objeto.
 
  

```python
class Animal:
    def __init__ (self, nome):
        self.nome = nome 

    def emitir_som(self):
        print('Som')

class Cachorro(Animal):
    def emitir_som(self):
        print("Au au")

class Gato(Animal):
    def emitir_som(self):
        print("Miau Miau")

lista = [
    Gato("Lia"),
    Cachorro("Rex")
]
for animais in lista:
    animais.emitir_som()
```

Requisitos: 
• A classe Veiculo deve ter placa, velocidade_max e velocidade_atual; torne os três atributos privados. 
• velocidade_atual começa em 0. 
• Crie getters para placa e velocidade_max. 
• Crie um setter para velocidade_max que aceite apenas valores positivos e nunca deixe a nova máxima menor que a velocidade atual. 
• acelerar(incremento=10) aumenta a velocidade sem ultrapassar a máxima. 
• frear(decremento=10) reduz a velocidade sem ficar negativa. 
• Implemente __str__ para mostrar placa, velocidade atual e máxima. 
• Demonstre o funcionamento com várias acelerações e frenagens. 

```python
class Veiculo:
    def __init__ (self,placa,velocidade_max):
        self.__placa = placa
        self.__velocidade_max = velocidade_max
        self.__velocidade_atual = 0

    def get_placa(self):
        return self.__placa

    def get_velocidade_max(self):
        return self.__velocidade_max

    def set_velocidade_max(self, valor):
        if valor > 0 and valor >= self._velocidade_atual:
            self.__velocidade_max = valor
            
        else: 
            return "Valor inválido"
        return self.__velocidade_max

    def acelerar(self,incremento = 10 ):
        self.__velocidade_atual += incremento
        
        if self.__velocidade_atual > self.__velocidade_max:
            self.__velocidade_max = self.__velocidade_atual

        return self.__velocidade_atual


    def frear (self, decremento = 10 ):  
        self._velocidade_atual -= decremento
        if self.__velocidade_atual < 0:
            self.__velocidade_atual = 0

        return self.__velocidade_atual
    

    def __str__(self):
        return f"Placa: {self.__placa} | Atual: {self.__velocidade_atual} | Máxima: {self.__velocidade_max}"


carro = Veiculo("ABC1234", 120)
print(carro)
print(carro.acelerar())
print(carro)
```

Exercício 8 — Funcionários e cálculo de bônus  [Média] 
Crie uma hierarquia para calcular bônus de funcionários de uma empresa. 
Requisitos: 
• Funcionario possui nome e salario; implemente __str__. 
• Funcionario.calcular_bonus() retorna 5% do salário. 
• Vendedor herda de Funcionario, acrescenta vendas e redefine calcular_bonus() para 5% do salário + 2% das vendas. 
• Gerente herda de Funcionario, acrescenta equipe e redefine calcular_bonus() para 10% do salário + R$ 100 por integrante da equipe. 
• Crie pelo menos um objeto de cada tipo, coloque-os em uma única lista e exiba nome, tipo e bônus usando o mesmo método calcular_bonus(). 


```python
class Funcionario:
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario= salario


    def calcular_bonus(self):
        bonus = self.salario * 0.05
        return bonus 

    def __str__(self):
        return f"Nome: {self.nome} | Salario: {self.salario} | Bonus: {self.calcular_bonus()}"

class Vendedor(Funcionario):
    def __init__(self, nome, salario, vendas):
        super().__init__(nome, salario)
        self.vendas = vendas

    def calcular_bonus(self):
        bonus_vendas = self.vendas * (2/100)
        return super().calcular_bonus() + bonus_vendas

class Gerente(Funcionario):
    def __init__(self, nome, salario, qtd_equipe):
        super().__init__(nome,salario)
        self.qtd_equipe = qtd_equipe


    def calcular_bonus(self):
        return (self.salario * 0.10) + (100 * self.qtd_equipe)


f1 = Funcionario("Julia", 1900)
v2 = Vendedor("Felipe", 4800,9)
g1 = Gerente("Roberta", 10000, 3)



funcionarios_gerais = [f1, v2, g1]

for funcionario in funcionarios_gerais:
    print(
    f"Nome: {funcionario.nome} | "
    f"Salário: {funcionario.salario} |  " 
    f"Tipo: {type(funcionario).__name__} | "
    f"Bônus: R$ {funcionario.calcular_bonus():.2f}"
    )


        
```

Exercício 9 — Produtos e contador de objetos  [Média] Crie uma classe Produto para um catálogo de e-commerce, explorando atributo de classe e encapsulamento.
 Requisitos: 
• Crie o atributo de classe contador iniciado em 0.
• Cada produto deve receber automaticamente um id sequencial a partir do contador. 
• Use atributos privados para id, nome e preco. 
• Crie getters para id, nome e preco. 
• Crie set_preco(novo_preco), aceitando apenas valores maiores que zero. 
• Implemente aplicar_desconto(percentual), validando percentual entre 0 e 100. 
• Implemente __str__. 
• Crie três produtos e, ao final, mostre quantos objetos foram criados usando Produto.contador. 

```python
class Produto:
    contador = 0 
    def __init__(self, nome, preco):
        Produto.contador += 1
        self._id = Produto.contador 
        self._nome = nome
        self._preco = preco



    def get_id(self):
        return self._id 

    def get_nome(self):
        return self._nome

    def get_preco(self):
        return self._preco

    def set_preco(self, novo_preco):
        if novo_preco > 0:
            self._preco = novo_preco

    def aplicar_desconto(self, percentual):
        if percentual > 0 and percentual < 100:
            desconto = self._preco * (percentual/100)
            self._preco = self._preco - desconto 
        return self._preco

    def __str__(self):
        return f'O ID do produto é {self._id} | O nome do produto é {self._nome} | O preço total do produto é R${self._preco:.2f}  '

p1 = Produto("Tomate", 10)
p2 = Produto("Cebola", 20)
p3 = Produto("Batata", 30)

produtos = [
    p1, p2 , p3
]

for produto in produtos:
    print (produto.get_id())
    print (produto.get_nome())
    print (produto.get_preco())
    print (produto.aplicar_desconto(10))



```

Exercício 10 — Mini sistema de pedidos de e-commerce  [Difícil] 
Desenvolva um pequeno sistema orientado a objetos para pedidos de um e-commerce. O objetivo é integrar a maior parte dos conceitos trabalhados na aula em um único problema. 
Requisitos: 
• Crie uma classe Produto com id automático (atributo de classe), nome e preço privados; inclua getters, setter validado para preço e __str__. 
• Crie ProdutoFisico herdando de Produto, com peso, e ProdutoDigital herdando de Produto, com tamanho_mb. 
• Em Produto, crie calcular_frete() retornando 0. Redefina-o em ProdutoFisico para retornar R$ 5,00 + R$ 2,00 por kg. ProdutoDigital deve manter frete igual a 0. Essa chamada comum deve demonstrar polimorfismo. 
• Crie uma classe Cliente com nome e email privados, getters e __str__.
• Crie uma classe Pedido com número automático (atributo de classe), cliente, lista de produtos e status privado. 

• Pedido deve oferecer adicionar_produto(produto), calcular_total(), calcular_frete_total(), fechar_pedido() e __str__. 
• Depois de fechado, o pedido não deve aceitar novos produtos. Essa regra deve ser controlada internamente pela classe. 
• calcular_total() deve somar os preços dos produtos. calcular_frete_total() deve chamar calcular_frete() de cada item sem testar se ele é físico ou digital. 
• No programa principal, crie um cliente, ao menos três produtos (misturando físicos e digitais), monte um pedido, exiba o resumo, feche-o e tente adicionar mais um produto para comprovar a regra. 

```python
class Produto:
    ID = 0 
    def __init__(self, nome, preco):
        Produto.ID += 1
        self._id = Produto.ID
        self._nome = nome
        self._preco = preco

    def get_preco (self):
        return self._preco

    def get_nome(self):
        return self._nome

    def set_preco(self, novo_valor):
        if novo_valor > 0:
            self._preco = novo_valor
        return self._preco

    def __str__(self):
        return f'O ID é {self._id} | O nome do produto é {self._nome} | O preço do produto é R${self._preco:.2f}'


    def calcular_frete(self):
        return 0 


class ProdutoFisico(Produto):
    def __init__(self, nome, preco, peso,):
        super().__init__(nome, preco)
        self.peso = peso

    def calcular_frete(self):
            valor_frete = 5 + (2 * self.peso)
            return valor_frete
    



class ProdutoDigital(Produto):
    def __init__(self, nome , preco, tamanho_mb):
        super().__init__(nome, preco)
        self.tamanho_mb = tamanho_mb


class Cliente:
    def __init__(self, nome, email):
        self._nome = nome
        self._email = email 


    def get_nome(self):
        return self._nome

    def get_email(self):
        return self._email

    def __str__(self):
        return f'O nome da cliente é {self._nome} e seu email é {self._email}'



class Pedido:
    cont = 0 
    def __init__(self, cliente):
        Pedido.cont += 1 
        self._id = Pedido.cont
        self._cliente = cliente
        self._lista_produtos = []
        self._status = "Aberto"

    def adicionar_produto(self , produto ):
        if self._status == "Aberto":
            self._lista_produtos.append(produto)
        else:
            return "O pedido ja está fechado"
        return self._lista_produtos
    
    def calcular_total(self):
        preco_total = 0 
        for produto in self._lista_produtos:
            preco = produto.get_preco()
            preco_total += preco
        return preco_total

    def calcular_frete_total(self):
        frete_total = 0

        for produto in self._lista_produtos:
            frete_total += produto.calcular_frete()
        return frete_total

    def fechar_pedido(self): 
        self._status = "Fechado"
        return self._status

    def __str__(self):
        texto = f"Pedido nº {self._id}\n"
        texto += f"Cliente: {self._cliente}\n"
        texto += f"Status: {self._status}\n"

        for produto in self._lista_produtos:
            texto += f"{produto}\n"

        texto += f"Total: R$ {self.calcular_total():.2f}\n"
        texto += f"Frete: R$ {self.calcular_frete_total():.2f}"

        return texto


  
cliente1 = Cliente("Julia", "julia@email.com")

p1 = ProdutoFisico("Notebook", 3500, 2)
p2 = ProdutoDigital("Curso de Python", 150, 500)
p3 = ProdutoFisico("Mouse", 80, 1)


pedido1 = Pedido(cliente1)

pedido1.adicionar_produto(p1)
pedido1.adicionar_produto(p2)
pedido1.adicionar_produto(p3)

print(pedido1)

pedido1.fechar_pedido()


p4 = ProdutoDigital("E-book", 50, 20)

pedido1.adicionar_produto(p4)

```
