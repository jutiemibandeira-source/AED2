```python
historicos = [
    ["ADS", "POO", "BD"],
    ["ADS", "ENG", "IA"],
    ["POO", "BD", "REDES"],
    ["ADS", "POO", "MOBILE"]
]

historico_aluno = ["ADS", "POO"]

def recomendar_disciplinas(historicos, historico_aluno):
    recomendacoes = []

    for outro_historico in historicos:

        possui_comum = False

        # verifica se existe disciplina em comum
        for disciplina in outro_historico:
            if disciplina in historico_aluno:
                possui_comum = True

        # adiciona recomendações
        if possui_comum:
            for disciplina in outro_historico:
                if disciplina not in historico_aluno and disciplina not in recomendacoes:
                    recomendacoes.append(disciplina)

    return recomendacoes


resultado = recomendar_disciplinas(historicos, historico_aluno)

print(resultado)
```
