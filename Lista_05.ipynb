5. (PRÁTICA): Classe Pessoa: 
Crie uma classe Pessoa com nome e idade. Adicione método 
faz_aniversario() que incrementa a idade em 1. Use __str__() para retornar algo como "<nome>, <idade> 
anos". Crie duas pessoas e mostre a idade de cada uma antes e depois de chamar faz_aniversario().

```python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def faz_aniversario(self):
        self.idade += 1

    def __str__(self):
        return f"{self.nome}, {self.idade} anos"


pessoa1 = Pessoa("Julia", 20)
pessoa2 = Pessoa("Ana", 25)

print("Antes do aniversário:")
print(pessoa1)
print(pessoa2)

pessoa1.faz_aniversario()
pessoa2.faz_aniversario()

print("\nDepois do aniversário:")
print(pessoa1)
print(pessoa2)
```

```python
class Lampada:
    def __init__(self):
        self.__estado = False  

    def ligar(self):
        self.__estado = True

    def desligar(self):
        self.__estado = False

    def is_acesa(self):
        return self.__estado

lampada = Lampada()

print(lampada.is_acesa()) 

lampada.ligar()
print(lampada.is_acesa())  

lampada.desligar()
print(lampada.is_acesa())
```

```python
class Animal:
    def fazer_som(self):
        print("Algum som")


class Cachorro(Animal):
    def fazer_som(self):
        print("Au Au")


class Gato(Animal):
    def fazer_som(self):
        print("Miau")


def animais_falam(lista_animais):
    for animal in lista_animais:
        animal.fazer_som()



cachorro = Cachorro()
gato = Gato()

lista = [cachorro, gato, Cachorro(), Gato()]

animais_falam(lista)
```
