```python
# Implementação da classe Fila

class Queue:

    # Inicia com uma fila vazia
    def __init__(self):
        self.itens = []

    # Verifica se fila está vazia
    def is_empty(self):
        return self.itens == []

    # Adiciona elemento no final da fila
    def enqueue(self, item):
        self.itens.insert(0, item)
      

    # Remove elemento do inicio da fila
    def dequeue(self):
        
        return self.itens.pop()

    # Retorna o número de elementos da fila
    def size(self):
        return len(self.itens)




def batata_quente(nomes, qtd_passadas):
    fila = Queue()

    for nome in nomes:
        fila.enqueue()

    while fila.size() > 1:
        for _ in range(qtd_passadas):
            fila.enqueue(fila.dequeue())

        fila.dequeue()

    vencedor = fila.dequeue()
        
    return vencedor

    
jogadores = batata_quente(['Julia','Ana', 'Mario','Felipe'],5)
print(f'O vencedor é {jogadores}')
```
