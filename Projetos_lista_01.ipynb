```python
'''Gerenciador de lista de contatos'''
from time import sleep as sl
nome_contatos = {}

def adicionar_contato(nome, telefone, email):
    if nome in nome_contatos:
        print(f"Contato '{nome}' já existe.")
    else:
        nome_contatos[nome] = {'telefone': telefone, 'email': email}
        print(f"Contato '{nome}' adicionado com sucesso.")

def buscar_contato(nome):
    if nome in nome_contatos:
        contato = nome_contatos[nome]
        print(f"Contato encontrado:")
        print(f"Nome: {nome}")
        print(f"Telefone: {contato['telefone']}")
        print(f"E-mail: {contato['email']}")

    else: 
        print(f"Contato '{nome}' não encontrado.")

def remover_contato(nome):
    if nome in nome_contatos:
        nome_contatos.pop(nome)
        print(f"Contato '{nome}' removido com sucesso.")

    else: 
        print(f"Contato '{nome}' não encontrado.")

def mostrar_contatos():
    if nome_contatos:
        print("Lista de contatos:")
        for nome, contato in nome_contatos.items():
            print(f"Nome: {nome}\nTelefone: {contato['telefone']}\nE-mail: {contato['email']}")
    else: 
        print("Nenhum usuário cadastrado.")
        
while True:
    opcao = input("Escolha uma opção (1 - Adicionar Contato, \n2 - Buscar Contato \n3 - Remover Contato \n4 - Mostrar Lista de Contatos \n5 - Sair)") 
    if opcao == "1":
        nome = input("Digite o nome do contato: ")
        telefone = input("Digite o telefone do contato: ")
        email = input("Digite o e-mail do contato: ")
        adicionar_contato(nome, telefone, email)

    elif opcao == "2":
        nome = input("Digite o nome do contato sdque deseja buscar: ")
        buscar_contato(nome)

    elif opcao == "3":
        nome = input("Digite o nome do contato que deseja remover: ")
        remover_contato(nome)

    elif opcao == "4":
        mostrar_contatos()

    elif opcao == "5":
        print("Saindo do programa...")
        sl(1)
        print ("Programa encerrado.")
        break

    else:
        print("Opção inválida. Tente novamente.")  

    
```

```python
'''Lista de Tarefas'''

from time import sleep as sl

tarefas = []

def adicionar_tarefa(tarefa):
    tarefas.append(tarefa)
    print("Tarefa adicionada com sucesso.")

def excluir_tarefa(indice):
    for i in range(len(tarefas)):
        if i == indice:
            tarefas.pop(i)
            print("Tarefa excluída com sucesso.")


def listar_tarefas():
    for i in range(len(tarefas)):
        print(f"{i} - {tarefas[i]}")

while True:
    opcao = input("Escolha uma opção: 1 - Adicionar Tarefa, 2 - Excluir Tarefa, 3 - Listar Tarefas, 4 - Sair: ")
    if opcao == "1":
        tarefa = input("Digite a tarefa:")
        adicionar_tarefa(tarefa)

    elif opcao == "2":
        listar_tarefas()
        indice = int(input("Digite qual indice da tarefa você deseja excluir:"))
        excluir_tarefa(indice)
 
    elif opcao == "3":
        listar_tarefas()

    elif opcao == "4":   
        print("Saindo do programa...")
        sl(1)
        print ("Programa encerrado.")
        break 
```
